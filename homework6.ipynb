```python
import numpy as np

arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(n):
    if n < 2:
        return False

    for divisor in range(2, int(n**0.5) + 1):
        if n % divisor == 0:
            return False

    return True

def prime_rows(arr):
    keep_rows = [any(is_prime(n) for n in row) for row in arr]
    return arr[keep_rows]

print(prime_rows(arr))
```

```python
import numpy as np

def checkerboard():
    board = np.zeros((8, 8), dtype=int)

    board[::2, ::2] = 1

    board[1::2, 1::2] = 1

    return board

print(checkerboard())
```

```python
def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)

    board[::2, 1::2] = 1
    board[1::2, ::2] = 1

    return board

print(reverse_checkerboard())
```

```python
import numpy as np

def expansion(universe, number):
    separator = " " * number
    return np.array([separator.join(word) for word in universe])

universe = np.array(["galaxy", "clusters"])

print(expansion(universe, 1))
print(expansion(universe, 2))
```

```python
import numpy as np

stars = np.array([
    [1123, 1456, 1789, 1324, 1876],
    [1567, 1987, 1678, 1405, 1589],
    [1345, 1654, 1523, 1109, 1923],
    [1298, 1890, 1367, 1784, 1432],
    [1823, 1756, 1489, 1672, 1550]
])

def secondDimmest(stars):
    sorted_stars = np.sort(stars, axis=1)
    return sorted_stars[:, -2]

print(secondDimmest(stars))
```
